### Dancing Night Parade

Source Pathfinder #168: King of the Mountain pg. 41

Level: 19

It has the Troop trait, meaning it represents a mass of spirits acting together, not a single creature. Mechanically, it behaves more like a moving crowd, swarm, or environmental event than a normal monster.

Its main gimmick is a festival/parade aura (“Riotous Parade”) that forces nearby creatures to laugh, dance, lose reactions, and struggle to concentrate.

It doesn’t even have a normal Strike routine like most monsters. Instead, it uses area effects and troop mechanics.

This thing is closer to:
